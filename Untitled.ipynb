```python
name_mark = list(map(str,input().strip().split()))
x = list(map(lambda x:int(x[1]) if x[0] % 2 !=0 else str(x[1]),enumerate(name_mark)))
y = [(x[i],x[i+1]) for i in range(0,len(x),2)] 
print(sorted(y,key=lambda x:x[1],reverse=True))
print(max(y,key=lambda x:x[1]))
print(min(y,key=lambda x:x[1]))
summ = 0
for i in y:
    summ += i[1]
avg = summ/len(y)
print(round(avg,2))
abv_avg = list(filter(lambda x:x[1]>avg,y))
print(abv_avg)
```

```python
# Read input
data = input().split()

# Build (name, marks) list
students = []
for i in range(0, len(data), 2):
    name = data[i]
    marks = int(data[i+1])
    students.append((name, marks))

# Sort by marks (descending)
students_sorted = sorted(students, key=lambda x: x[1], reverse=True)

# Topper (max using key)
topper = max(students, key=lambda x: x[1])

# Lowest scorer (min using key)
lowest = min(students, key=lambda x: x[1])

# Average marks
avg = sum(mark for _, mark in students) / len(students)

# Students above average
above_avg = [s for s in students if s[1] > avg]

# Output
print(students_sorted)
print(topper)
print(lowest)
print(above_avg)
```
