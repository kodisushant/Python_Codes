```python
#NORMAL DISTRIBUTION
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
np.random.seed(42)
marks = np.random.normal(70, 10, 15)
sort_marks = sorted(marks)
print(marks)
print(sort_marks)

plt.subplot(1,2,1)
plt.plot(marks)
plt.subplot(1,2,2)
plt.hist(marks)
plt.show()

sns.displot(marks,kde=True,bins=10)
```
