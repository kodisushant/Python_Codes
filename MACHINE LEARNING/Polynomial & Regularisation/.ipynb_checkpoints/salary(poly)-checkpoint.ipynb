```python
for d in [1, 2, 3, 4]:
    poly = PolynomialFeatures(degree=d)
    X_poly = poly.fit_transform(x)
    model = LinearRegression().fit(X_poly, y)
    plt.plot(X_grid, model.predict(poly.transform(X_grid)), label=f'degree {d}')

plt.scatter(x, y, color='red')
plt.legend()
plt.show()
```

```python
for d in range(1, 8):
    poly = PolynomialFeatures(degree=d)
    X_train_poly = poly.fit_transform(x_train)
    X_test_poly = poly.transform(x_test)
    model = LinearRegression().fit(X_train_poly, y_train)
    train_err = r2_score(y_train, model.predict(X_train_poly))
    test_err = r2_score(y_test, model.predict(X_test_poly))
    print(f"degree={d}, train_mse={train_err:.2f}, test_mse={test_err:.2f}")
```
