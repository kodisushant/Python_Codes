```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
df = pd.read_csv('loves_cool_as_ice.csv')

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
for col in ['loves_popcorn','loves_soda']:
    df[col] = le.fit_transform(df[col])
print(df.head())
X = df.drop('loves_cool_as_ice', axis=1)  
y = df['loves_cool_as_ice'] 
# X = df.drop(columns=['loves_cool_as_ice'])
# y = df['loves_cool_as_ice']



dt = DecisionTreeClassifier(criterion='gini', max_depth=3, random_state=42)
dt.fit(X, y)

# Visualize tree WITH gini values shown at each node
plt.figure(figsize=(14, 8))
plot_tree(dt, feature_names=X.columns, class_names=['no', 'yes'], filled=True, rounded=True)
plt.show()
```
