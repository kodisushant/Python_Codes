```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 
from sklearn.datasets import load_wine
```

```python
wine = load_wine(as_frame=True) 
```

```python
x = wine.data 
y = wine.target
```

```python
from sklearn.model_selection import train_test_split 
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier , plot_tree
dt_model = DecisionTreeClassifier(criterion='gini',max_depth=4,random_state=42)
dt_model.fit(x_train,y_train) 
```

```python
y_pred1 = dt_model.predict(x_test)
```

```python
from sklearn.metrics import classification_report , accuracy_score , precision_score
print("Accuracy:", accuracy_score(y_test, y_pred1))
print("Precision:", precision_score(y_test, y_pred1,average='weighted'))
print(classification_report(y_test, y_pred1, target_names=wine.target_names))
```

```python
from sklearn.model_selection import cross_val_score
depths = range(1,15)
scores = [] 

for depth in depths: 
    Tree = DecisionTreeClassifier(max_depth=depth ,criterion='gini',random_state=42)
    cv_scores = cross_val_score(Tree , x_train,y_train , cv=5)
    scores.append(cv_scores.mean())

plt.plot(depths,scores,marker='o')
plt.xlabel("max depth")
plt.ylabel("cv accuracy")
plt.title("Finding optimal max_depth")
plt.show() 

bestDepth = depths[np.argmax(scores)]
print("Best max_depth:", bestDepth) 
```

```python
plt.figure(figsize=(10,10))
plot_tree(dt_model,feature_names=wine.feature_names , class_names= wine.target_names , filled=True)
plt.show()
```

## RANDOM FOREST

```python
from sklearn.ensemble import RandomForestClassifier 
rf_model = RandomForestClassifier(n_estimators=100,max_depth=4,random_state=42)
rf_model.fit(x_train,y_train)
```

```python
y_pred2 = rf_model.predict(x_test)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred2))
print(classification_report(y_test, y_pred2 , target_names=wine.target_names))

# Feature importance (averaged across all trees)
for name, importance in zip(wine.feature_names, rf_model.feature_importances_):
    print(f"{name}: {importance:.3f}")
```
