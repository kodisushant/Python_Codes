```python
import pandas as pd
import numpy as np
data={
    'name':['ravi','priya','arjun','meena','kiran'],
    'marks':[45,78,92,60,88],
    'sh':[2,5,8,4,7]
}
df=pd.DataFrame(data)
print(df['marks'].mean())
print(df['marks'].std())
df['marks']=(df['marks']-df['marks'].mean())/df['marks'].std()
df['marks']=np.round((df['marks']-df['marks'].min())/(df['marks'].max()-df['marks'].min()),2)
df['sh']=(df['sh']-df['sh'].mean())/df['sh'].std()
df['sh']=np.round((df['sh']-df['sh'].min())/(df['sh'].max()-df['sh'].min()),2)
print(df)
```

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler
scaler=MinMaxScaler()
df['marks_norm_sklearn']=scaler.fit_transform(df[['marks']])
scaler2=StandardScaler()
df['marks_std_sklearn']=scaler2.fit_transform(df[['marks']])
```
