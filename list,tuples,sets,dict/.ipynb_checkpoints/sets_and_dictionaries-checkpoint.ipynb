```python
''' 
Q5. Given data = [5, 3, 8, 3, 9, 5, 1, 8, 2], write a program to:
(a) Print all unique values
(b) Print the count of duplicate entries that were removed
(c) Print the unique values in sorted order
'''
data = [5, 3, 8, 3, 9, 5, 1, 8, 2] 
d = list(set(data))
print(d)
count = 0
for i in d:
    if data.count(i) > 1: 
        count+=1
print(count)
```

```python
''' 
Q17. Three friends each have a set of movies they have watched:

alice = {'Inception', 'Interstellar', 'Dune', 'Oppenheimer'}
bob   = {'Dune', 'Tenet', 'Inception', 'Avatar'}
carol = {'Interstellar', 'Dune', 'Gravity', 'Inception'}
Find:
(a) Movies all three have watched ( inception , dune  )
(b) Movies watched by Alice and Bob but not Carol ( none )
(c) All unique movies across all three ( gravity , avatar tenet openhiemer )
'''

alice = {'Inception', 'Interstellar', 'Dune', 'Oppenheimer'}
bob   = {'Dune', 'Tenet', 'Inception', 'Avatar'}
carol = {'Interstellar', 'Dune', 'Gravity', 'Inception'} 

all_3_watc = alice.intersection(bob).intersection(carol)
print("Movies watched by all three are ",all_3_watc)

only_al_bb = alice.intersection(bob) - all_3_watc
print("Movies watched by only both alice and bob are",only_al_bb)

unq = alice.union(bob).union(carol) - alice.intersection(bob)- alice.intersection(carol)- carol.intersection(bob) - all_3_watc
print("Unique movies are",unq)
```

```python
''' 
Q23. Given:

inventory = {'pen': 50, 'book': 30, 'eraser': 80, 'ruler': 20}
Write code to:
(a) Find the item with highest stock
(b) Find the item with lowest stock
(c) Print total stock count
 
'''
inventory = {'pen': 50, 'book': 30, 'eraser': 80, 'ruler': 20}
print(f"The highest stock in the inventory is {max(inventory,key=lambda x:inventory[x])}")
print(f"The lowest stock in the inventory is {min(inventory,key=lambda x:inventory[x])}")
print(f"the total Stock count is {sum(inventory.values())}")
```

```python
''' 
Q25. A vote counting program collects votes as a list:

votes = ['Alice', 'Bob', 'Alice', 'Carol', 'Bob', 'Alice', 'Carol', 'Bob', 'Alice']
You solved this before using count(). Now redo it using a dictionary.
Print each candidate's vote count and declare the winner.
'''
votes = ['Alice', 'Bob', 'Alice', 'Carol', 'Bob', 'Alice', 'Carol', 'Bob', 'Alice']
freq = {}
for i in votes: 
    freq[i] = freq.get(i,0)+1
for k,v in freq.items(): 
    print(f"{k:<7}: {v} votes")
print("The winner is",max(freq,key=lambda x:freq[x]))
```

```python
''' 
Q30. Given:

students = {
    'Ravi':  {'marks': 72, 'branch': 'CSE'},
    'Priya': {'marks': 91, 'branch': 'ECE'},
    'Arjun': {'marks': 65, 'branch': 'CSE'},
    'Meena': {'marks': 85, 'branch': 'CSE'}
}
Write code to:
(a) Print names of all CSE students
(b) Find the CSE student with highest marks
(c) Print average marks of all students
 
'''
students = {
    'Ravi':  {'marks': 72, 'branch': 'CSE'},
    'Priya': {'marks': 91, 'branch': 'ECE'},
    'Arjun': {'marks': 65, 'branch': 'CSE'},
    'Meena': {'marks': 85, 'branch': 'CSE'}
} 
list(map(lambda x:print(x[0]) if x[1]["branch"]=="CSE" else None ,students.items()))
print("The student with highest marks is",max(students.items() ,key=lambda x:x[1]["marks"])[0])
avg = sum(list(map(lambda x : x[1]["marks"],students.items())))/len(students)
print(f"The average of all students is {avg:.2f}")
```

```python
# method 2 for above question 
students = {
    'Ravi':  {'marks': 72, 'branch': 'CSE'},
    'Priya': {'marks': 91, 'branch': 'ECE'},
    'Arjun': {'marks': 65, 'branch': 'CSE'},
    'Meena': {'marks': 85, 'branch': 'CSE'}
}  
for name , info in students.items():
    if info["branch"] == "CSE": 
        print(name)
print("The student with highest marks is",max(students.items() ,key=lambda x:x[1]["marks"])[0])
avg = sum(info["marks"] for info in students.values())/len(students)
print(f"The average of all students is {avg:.2f}")
```

```python
''' 
Q37. Given text = "she sells seashells by the seashore", write a program to:
(a) Find count of unique words (use a set)
(b) Find frequency of each word (use a dictionary)
(c) Find the most repeated word
'''
text = "she sells seashells by the seashore" 
unq = set(text)
freq = {i:text.count(i) for i in unq}
print(freq)
print("The most repeated letter is",max(freq.items(),key=lambda x:x[1])[0])
```

```python
''' 
Q38. You have two lists:

names  = ['Ravi', 'Priya', 'Arjun', 'Meena']
scores = [72, 91, 65, 85]
Use zip() to create a dictionary pairing each name with their score. Then find the topper using max() with key=.

'''
names  = ['Ravi', 'Priya', 'Arjun', 'Meena']
scores = [72, 91, 65, 85] 
h = zip(names,scores)
d = dict(h)
print(d)
print("The topper is",max(d.items(),key=lambda x:x[1])[0])
```

```python
''' 
Q39. Given:

roster_a = ['Ravi', 'Priya', 'Arjun', 'Meena']
roster_b = ['Priya', 'Kiran', 'Arjun', 'Sneha']
Using sets, find students common to both sections. 
Then create a dictionary showing each common student's section count (value should be 2 for all of them).

'''
roster_a = ['Ravi', 'Priya', 'Arjun', 'Meena']
roster_b = ['Priya', 'Kiran', 'Arjun', 'Sneha'] 
ra = set(roster_a)
rb = set(roster_b)
alll = ra.union(rb)
common = ra.intersection(rb)
sec_count = {i:2 if i in common else 1 for i in alll}
sec_count
```

```python
''' 
Q40. A library system stores borrowed books per student:

library = {
    'Ravi':  ['Python Basics', 'DSA', 'DBMS'],
    'Priya': ['OS', 'Python Basics', 'Networks'],
    'Arjun': ['DSA', 'OS', 'Python Basics']
}
Write a program to find:
(a) All unique book titles across all students (use a set)
(b) Which book is borrowed by the most students
(c) Students who have borrowed 'DSA'
 
'''
library = {
    'Ravi':  ['Python Basics', 'DSA', 'DBMS'],
    'Priya': ['OS', 'Python Basics', 'Networks'],
    'Arjun': ['DSA', 'OS', 'Python Basics']
} 
unq = set(i  for books in library.values() for i in books)
print(unq)
freq = {book: sum(1 for books in library.values() if book in books) for book in unq} 
print("most borrowed books by students is",max(freq.items(),key=lambda x:x[1])[0])
borrow_dsa = [k for k,v in library.items()
              if "DSA" in v ]
print(borrow_dsa)
```
