```python
employees = [
    {'name': 'Ravi',  'dept': 'IT',  'salary': 75000},
    {'name': 'Priya', 'dept': 'HR',  'salary': 55000},
    {'name': 'Arjun', 'dept': 'IT',  'salary': 90000},
    {'name': 'Meena', 'dept': 'HR',  'salary': 60000},
    {'name': 'Kiran', 'dept': 'IT',  'salary': 80000},
    {'name': 'Sita',  'dept': 'Fin', 'salary': 95000},
    {'name': 'Ram',   'dept': 'Fin', 'salary': 85000},
]
depts = set(emp["dept"] for emp in employees)

def avg_sal(dept):
    sal = [emp["salary"] for emp in employees if emp["dept"]==dept]
    return sum(sal)/len(sal)
print("Highest Salary Employee:",max(depts,key=avg_sal))
```
