```python
def sum_multiples(n, k):
    m = (n - 1) // k
    return k * m * (m + 1) // 2

T = int(input())
for _ in range(T):
    N = int(input())
    print(sum_multiples(N, 3) + sum_multiples(N, 5) - sum_multiples(N, 15))
```

```python
''' 
Sample Input

3
Krishna 67 68 69
Arjun 70 98 63
Malika 52 56 60
Malika

Sample Output
56.00
'''
n = int(input())
student_marks = {}
for _ in range(n):
    name, *line = input().split()
    scores = list(map(float, line))
    student_marks[name] = scores
query_name = input()
summ = sum(student_marks[query_name])
avg = summ/len(student_marks[query_name])
print(f"{avg:.2f}")
```

```python
# Enter your code here. Read input from STDIN. Print output to STDOUT
import numpy as np
n,m = map(int,input().split())

a = np.array( [list(map(int,input().split())) for _ in range(n)] )
b = np.array( [list(map(int,input().split())) for _ in range(n)] )
print(np.add(a,b))
print(np.subtract(a,b))
print(np.multiply(a,b))
print(np.int.divide(a,b))
print(np.mod(a,b))
print(np.power(a,b))
```

```python
def minion_game(st):
    # your code goes here
    stuart_score = 0
    stuart = []
    kev = []
    kev_score = 0
    v = ["A","E","I","O","U"]
    for i in range(len(st)):
        for j in range(i+1,len(st)+1):
            if st[i:j][0] not in v:
                stuart.append(st[i:j])
                stuart_score+=1
            else:
                kev.append(st[i:j])
                kev_score+=1
    # print(stuart)  
    print(stuart_score)
    # print(kev) 
    print(kev_score)
        

if __name__ == '__main__':
    s = input()
    minion_game(s)
```
