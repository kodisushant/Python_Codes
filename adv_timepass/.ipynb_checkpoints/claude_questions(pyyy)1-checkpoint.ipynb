```python
text = "the cat sat on the mat the cat wore a hat"
words = text.split()
print(words)
# Your task:
# Write a function is_repeated(word) that returns True if the word appears more than once.
# Then use a set comprehension to collect all such repeated words. 

def is_repeated(word): 
    return (words.count(word)>1)

repeated = set(w for w in words if is_repeated(w))
print(repeated)

repeated2 = set(filter(lambda x:words.count(x)>1,words))
repeated2
```

```python
students = [
    {'name': 'Aman',  'score': 88},
    {'name': 'Bina',  'score': 45},
    {'name': 'Charu', 'score': 73},
    {'name': 'Dev',   'score': 59},
    {'name': 'Esha',  'score': 92},
]

# Your task:
# Write a function grade(score) that returns 'A' (>=80), 'B' (>=60), or 'C' (below 60).
# Then build a new list of dicts: [{'name': ..., 'grade': ...}, ...]
# using a list comprehension that calls your function.

def grade(score): 
    if score >=80: 
        return 'A'
    elif score >=60: 
        return 'B'
    else: 
        return 'C'
new_student = [ {"name":i["name"] , "GRade" : grade(i["score"])} for i in students ]


print(new_student)
```
