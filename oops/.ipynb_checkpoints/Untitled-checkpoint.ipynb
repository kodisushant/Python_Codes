```python
class pokemon:
    def __init__(self,name,atk,dfn,sp_a,sp_d,hp,speed,moves): 
        self.name=name
        self.attack=atk 
        self.defence=dfn 
        self.special_a=sp_a
        self.special_d=sp_d
        self.health = hp 
        self.speed=speed 
        self.moves = moves
    def attacks(self): 
        for i,j in enumerate(self.moves):
            print(i+1,j)
        num = int(input("Select the move number:"))
        return f"{self.name} used {self.moves[num-1]}"
```

```python
pikachu = pokemon("Pikachu",55,40,50,50,35,90,["Thunderbolt","Thunder"])
print(pikachu.speed)
print(pikachu.moves)
pikachu.attacks()
```

```python
class Superhero:
    def __init__(self,name):
        self.name = name

class Experience:
    @staticmethod
    def expertise(exp): 
        if exp<=25: 
            return "Beginner"
        elif exp<=50: 
            return "Hero"
        elif exp<=100:
            return "GrandMaster"


class Attack(Experience): 
    base_atk = 25
    def damage(self,exp): 
        level = Attack.expertise(exp)
        if level == "Beginner": 
            return self.base_atk*1
        elif level == "Hero": 
            return self.base_atk*2
        elif level == "GrandMaster": 
            return self.base_atk*4


class Fire(Superhero,Attack):
    def __init__(self,name,exp):
        super().__init__(name)
        self.Exp = exp 
        self.Power = "Fire"
        self.Moves = {"Flamethrower":10,"Fire Tiger":15}
    def attacks(self): 
        for i,j in enumerate(self.Moves):
            print(i+1,j)
        num = int(input("Select the move number:"))
        move = list(self.Moves.keys())[num-1]
        print(f"{self.name} used {move}")
        print("Damage :",self.damage(self.Exp)+ self.Moves[move] )


class Water(Superhero,Attack):
    def __init__(self,name,exp):
        super().__init__(name)
        self.Exp = exp 
        self.Power = "Water"
        self.Moves = {"Hydro Pump":10,"Whale Splash":25}
    def attacks(self): 
        for i,j in enumerate(self.Moves):
            print(i+1,j)
        num = int(input("Select the move number:"))
        move = list(self.Moves.keys())[num-1]
        print(f"{self.name} used {move}")
        print("Damage :",self.damage(self.Exp)+ self.Moves[move] )


class Thunder(Superhero,Attack):
    def __init__(self,name,exp):
        super().__init__(name)
        self.Exp = exp 
        self.Power = "Thunder"
        self.Moves = {"Lightning Speed":15,"Thunder":20}
    def attacks(self): 
        for i,j in enumerate(self.Moves):
            print(i+1,j)
        num = int(input("Select the move number:"))
        move = list(self.Moves.keys())[num-1]
        print(f"{self.name} used {move}")
        print("Damage :",self.damage(self.Exp)+ self.Moves[move] )

def about(hero):
    print("Name:",hero.name)
    print("Powers:",hero.Power)
    print("Moves:",hero.Moves)
    print("Exp Points:",hero.Exp)
    level = hero.expertise(hero.Exp)
    print("Rank:",level)
    

agni = Fire("Agni",10)
# agni.attacks() 
sun = Fire("Sun",50)
# sun.attacks() 

varun = Water("Varun",90)
# varun.attacks() 

thor = Thunder("Thor",85)
# thor.attacks() 

about(agni)
agni.Exp += 40
about(agni)
agni.Moves["Fire Dragon"] = 25
about(agni)

varun.Moves["Kraken orchestration"] = 35
about(varun)
```
