```python
marks = np.array([
    45, 52, 67, 71, 88, 90, 56, 62, 74, 81,
    95, 49, 53, 68, 77, 84, 91, 59, 64, 72,
    85, 93, 47, 58, 69, 76, 82, 89, 96, 54
])
plt.hist(marks,bins=5)
plt.xlabel("Marks")
plt.ylabel("No of Students")
plt.title("std vs marks")
```

```python
products = np.array(["Laptop", "Phone", "Tablet", "Watch", "Headphones"])
sales = np.array([120, 250, 90, 180, 150])

plt.bar(products,sales)
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Units Sold")
```

```python
companies = np.array(["Apple", "Samsung", "Xiaomi", "OnePlus"])
share = np.array([35, 30, 20, 15])
explode = [0.1, 0, 0, 0]
plt.pie(share,labels=companies,autopct="%1.1f%%",explode=explode,shadow=True)
plt.title("Smartphone Market share")
```
