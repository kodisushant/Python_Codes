```python
import numpy as np 
import matplotlib.pyplot as plt 
np.random.seed(7) 

#Stimulate marks for 3 subjects - diff distribution 
maths = np.random.normal(72,12,80)
python = np.random.normal(80,8,80)
dsa = np.random.normal(68,15,80)

maths[2] = 15 
python[3] = 99
dsa[4] = 105 
dsa[0] = 10 

maths = np.clip(maths,0,100)
python = np.clip(python,0,100)
dsa = np.clip(dsa,0,100) 

fig , ax = plt.subplots(figsize=(9,6))

bp = ax.boxplot(
    [maths,python,dsa], 
    label=["maths","Python","dsa"], 
    patch_artist=True , 
    notch=False , 
    showmeans=True , 
    meanprops=dict(marker="D",markerfacecolor="yellow",markeredgecolor="black",markersize=8 ), 
    boxprops = dict(linewidth=1.5), 
    medianprops=dict(color="red",linewidth=2.5), 
    whiskerprops=dict(linestyle="--",linewidth=1.2), 
    flierprops=dict(marker="o",markerfacecolor="red",markersize=6,alpha=0.7),
    # color=("Blue","Red","Green")
)
# c = ["Red","Blue","Green"]
ax.set_title("Class performance")
ax.set_xlabel("Subject")
ax.set_ylabel("Marks")
ax.legend()
```
