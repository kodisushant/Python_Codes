```python
branches = ['CSE', 'AI&ML', 'ECE', 'MECH', 'CIVIL']
students = [240, 180, 150, 120, 90]
colors   = ['#4285F4','#EA4335','#34A853','#FBBC05','#9C27B0']
explode  = (0.05, 0.05, 0, 0, 0)  # highlight top 2
fig, ax = plt.subplots(figsize=(7, 6))
ax.pie(students, labels=branches, colors=colors,
       autopct='%1.1f%%', startangle=140,
       wedgeprops=dict(width=0.5))  # width < 1 = donut
ax.set_title('KMIT - Branch Distribution (Donut Chart)')
plt.tight_layout()
plt.savefig('donut_chart.png', dpi=150)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 7))
wedges, texts, autotexts = ax.pie(
    students,
    labels=branches,
    colors=colors,
    explode=explode,
    autopct='%1.1f%%',
    startangle=140,
    shadow=True,
    pctdistance=0.80
)

# Style the percentage text
for autotext in autotexts:
    autotext.set_fontsize(10)
    autotext.set_fontweight('bold')

ax.set_title('KMIT - Branch-wise Student Distribution\n'
             'Total Students: 780',
             fontsize=13, pad=20)
plt.tight_layout()
plt.savefig('pie_branch.png', dpi=150)
plt.show()
```
