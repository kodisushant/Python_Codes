```python
import matplotlib.pyplot as plt 
import numpy as np 
test = np.array([1,2,3,4,5])
ravi = np.array([45,52,60,68,75])
priya = np.array([70,68,72,75,80])
arjun = np.array([55,50,48,60,65])

plt.figure(figsize=(9,5))
plt.plot(test,ravi,"b-o",linewidth=2,markersize="7",label="ravi")
plt.plot(test,priya,"r-s",linewidth=2,markersize="7",label="Priya") 
plt.plot(test,arjun,"g-*",linewidth=2,markersize="7",label="arjun")
plt.axhline(y=50,color="black",linestyle="--",linewidth=1.2,label="Pass Mark (50")
plt.title("KMIT STUDENT PERFORMANCE ACROSS ")
plt.xlabel("Exam number")
plt.ylabel("Marks")
plt.xticks(tests) 
plt.ylim(30, 90)
plt.legend(loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout() 
plt.savefig('line_plot_performance.png', dpi=150)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

# Study hours vs marks for 80 KMIT students
study_hours = np.random.uniform(1, 8, 80)
marks = 40 + 7 * study_hours + np.random.normal(0, 8, 80)
marks = np.clip(marks, 0, 100)
branch = np.random.choice(['CSE','AI&ML','ECE'], 80)

# --- Basic scatter with color by branch ---
color_map = {'CSE': 'blue', 'AI&ML': 'red', 'ECE': 'green'}
colors    = [color_map[b] for b in branch]

fig, ax = plt.subplots(figsize=(9, 6))
for br, col in color_map.items():
    mask = branch == br
    ax.scatter(study_hours[mask], marks[mask],
               c=col, s=60, alpha=0.7,
               edgecolors='black', linewidths=0.5,
               label=br)

# Compute and show correlation
r = np.corrcoef(study_hours, marks)[0, 1]
ax.set_title(f'KMIT - Study Hours vs Exam Marks\n'
             f'Correlation r = {r:.2f}', fontsize=12)
ax.set_xlabel('Daily Study Hours')
ax.set_ylabel('Exam Marks')
ax.legend(title='Branch')
ax.grid(True, linestyle='--', alpha=0.4)

# Add trend line
m, b = np.polyfit(study_hours, marks, 1)
x_line = np.linspace(1, 8, 100)
ax.plot(x_line, m*x_line + b, 'k--',
        linewidth=1.5, label=f'Trend: y={m:.1f}x+{b:.1f}')
ax.legend(title='Branch')
plt.tight_layout()
plt.savefig('scatter_study_marks.png', dpi=150)
plt.show()

# --- Bubble chart (scatter with variable size) ---
np.random.seed(10)
attendance = np.random.uniform(60, 100, 50)
marks2     = 30 + 0.6 * attendance + np.random.normal(0,5,50)
assignments = np.random.randint(1, 10, 50)  # bubble size

fig, ax = plt.subplots(figsize=(8, 6))
sc = ax.scatter(attendance, marks2,
                s=assignments * 30,
                c=marks2, cmap='RdYlGn',
                alpha=0.7, edgecolors='black')
plt.colorbar(sc, label='Marks')
ax.set_title('Attendance vs Marks\n'
             '(Bubble size = Assignments submitted)')
ax.set_xlabel('Attendance (%)')
ax.set_ylabel('Marks')
ax.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.savefig('bubble_chart.png', dpi=150)
plt.show()
```
