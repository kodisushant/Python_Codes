```python
''' 
Q13. Write a function char_frequency(s, ch) using only indexing, len(), and a loop (no .count())
that returns how many times ch appears in s.
Expected: char_frequency("programming", 'g') → 2
'''
def char_freq(s,ch ):
    count = 0
    for i in s: 
        if i == ch:
            count+=1
    return count
c= char_freq("programming", 'g') 
print(c)
```

```python
''' 
Q15. Write a function rotate(s, n) that rotates the string n positions to the left using only slicing.
Expected: rotate("abcdef", 2) → 'cdefab'; rotate("abcdef", -1) → 'fabcde' 
'''
def rotate(s,n):
    n=n%len(s)
    return s[n:]+s[:n]
word = input("Enter a word to rotate:")
num = int(input("Enter number of times to rotate:"))
rotated = rotate(word,num)
print(rotated)
```

```python
''' 
Q18. Given email = "  User.Name@Example.COM  ", write code to normalise it: strip whitespace,
convert to lowercase. Then check if it ends with .com and starts with user.
Expected: 'user.name@example.com', True, True
'''
email = "  User.Name@Example.COM  "
normal_email = email.strip().lower()
print(normal_email)
# print(normal_email.rsplit(".",1)[1]) 
print("True") if normal_email.rsplit(".",1)[1] == "com" else print("false")
print("true") if normal_email.split(".",1)[0] == "user" else print("false")
```

```python
''' 
Q20. Given data = "name:Alice;age:30;city:Delhi", split by ';' to get key-value pairs,
then split each pair by ':'. Build a dictionary from the result.
Expected: {'name': 'Alice', 'age': '30', 'city': 'Delhi'}
'''
data = "name:Alice;age:30;city:Delhi"
fist_split = data.split(";")
sec_split = list(map(lambda x:x.split(":"),fist_split))
d = {i[0]:i[1] for i in sec_split }
print(d)

#method 2 
d2 = {k:v for k,v in (i.split(":") for i in data.split(";"))}
print(d2)
```

```python
''' 
Q23. Write a function is_palindrome_word(word) that strips whitespace, converts to lowercase,
and checks if the result equals its reverse using slicing.
Expected: is_palindrome_word("  Racecar  ") → True; is_palindrome_word("hello") → False
 
'''
def is_palindrome_word(word):
    return word.strip().lower() == word.strip().lower()[::-1]
print(is_palindrome_word("  Racecar  "))
print(is_palindrome_word("hello"))
```

```python
''' 
Q25. Given phone = "(040) 1234-5678", use chained .replace() 
calls to strip all non-digit characters (, ), -,   and produce a pure digit string.
Expected: '04012345678'
'''
import re
phone = "(040) 1234-5678"
new_phone = re.sub(r"[\s()-]","",phone)
print(new_phone)

#method 2
clean = "".join(filter(str.isdigit, phone))
print(clean)

#method 3 
phone = "(040) 1234-5678"

clean = phone.replace("(", "").replace(")", "").replace(" ", "").replace("-", "")
print(clean)
```

```python
''' 
Q26. Write a function title_clean(s) that: strips whitespace, converts to title case,
removes duplicate spaces inside the string (replace '  ' with ' ' in a loop until no doubles remain),
and returns the result.
Expected: title_clean("  hello    world  ") → 'Hello World'
'''
def title_clean(s): 
    new = s.strip().title()
    for _ in new: 
        if new.count("  ") >=1: 
            new = new.replace("  "," ")
            continue
        else: 
            break
    return new
print(title_clean("  hello    world  "))
```

```python
''' 
Q27. Given a log line log = "[ERROR] 2024-01-15 :: Disk full on /dev/sda1", 
use .split() and string methods to extract: 
(a) the log level (ERROR),
(b) the date (2024-01-15), 
(c) the message after '::'.
Expected: 'ERROR', '2024-01-15', ' Disk full on /dev/sda1'
'''
log = "[ERROR] 2024-01-15 :: Disk full on /dev/sda1" 
print(log.split()[0])
print(log.split()[1])
print(log.split("::")[1])
```

```python
''' 
Q28. Given tags = "#python #data #ml #ai", split by space to get a list of tags,
then strip the # from each tag using .lstrip('#') and rejoin with ', '.
Expected: 'python, data, ml, ai'
'''
tags = "#python #data #ml #ai" 
tag = tags.split()
sub = list(map(lambda x : x.lstrip("#"),tag))
print(",".join(sub))
```

```python
''' 
Q29. Write a function word_count(sentence) that: lowercases the input, removes all punctuation (.,!?;:),
splits into words, and returns a dictionary of word → frequency.
Expected: word_count("To be, or not to be!") → {'to': 2, 'be': 2, 'or': 1, 'not': 1}
'''
def word_count(sentence):
    punctuation = [".",",","!","?",";",":"]
    clean = sentence.strip().lower().split()
    for i in range(len(clean)):
        if punctuation in clean[i]:
            clean[i] = clean[i].strip(punctuation)
            
```
