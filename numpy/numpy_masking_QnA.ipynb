```python
'''
Exercise 15 — Normalisation
Given `data = [10, 85, 42, 67, 5, 91, 33]`, use boolean masking to:
 Replace values below 20 with 20 (lower clamp)
 Replace values above 80 with 80 (upper clamp)
'''
data = np.array([10, 85, 42, 67, 5, 91, 33])
data[data<20] = 20
data[data>80] = 80 
print(data)
```

```python
data = np.array([10, 85, 42, 67, 5, 91, 33])
data = np.where(data<20,20,data)
data = np.where(data>80,80,data)
print(data)
```

```python
''' 
Bonus Challenge

Exercise 16
Given a 2D marks array:
python
marks = np.array([[55, 74, 43],
                  [90, 61, 38],
                  [47, 82, 74]])

 Find all values below 50 and replace them with 50 (pass mark)
 Count how many values were below 50 originally


'''
marks = np.array([[55, 74, 43],
                  [90, 61, 38],
                  [47, 82, 74]])
c_50 = np.sum(marks == 50)
marks[marks<50] = 50
less_than_50 = np.sum(marks == 50)-c_50
print(less_than_50)
```

```python
''' BETTER APPROACH FOR ABOVE '''
marks = np.array([[55, 74, 43],
                  [90, 61, 38],
                  [47, 82, 74]])
less_than_50 = np.sum(marks<50)
marks[marks<50] = 50
print(marks)
print(less_than_50)
```
