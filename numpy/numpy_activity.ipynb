```python
'''Use the -1 shortcut in reshape to avoid manually computing one dimension. Rewrite each reshape using -1:

a = np.arange(20)
a.reshape(4, 5)   # rewrite using -1 for the second dim
a.reshape(5, 4)   # rewrite using -1 for the first dim
a.reshape(2, 2, 5)# rewrite using -1 for the last dim'''
import numpy as np

a = np.arange(20)

a.reshape(4, -1)     # (4, 5)

a.reshape(-1, 4)     # (5, 4)

a.reshape(2, 2, -1)  # (2, 2, 5) 
```
