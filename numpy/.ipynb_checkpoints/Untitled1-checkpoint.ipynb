```python
import numpy as np
np.random.seed(1)
names = np.array(["a","b","c","d"])
cgpa = np.array([np.random.random()*10 for _ in range(len(names))])
cgpa = np.round(cgpa,2)
print(cgpa,names)
ind = np.argsort(cgpa)
print(np.sort(cgpa))
print(names[ind])
# print(names)
top = names[ind]
print(top)
```

```python
# Explicit dtype
import numpy as np
a = np.array([1, 2, 3], dtype=np.float64)
print(a.dtype)      # float64
print(a)
# Type conversion
b = a.astype(np.int32)
print(b.dtype)      # int32
print(b)
print(type(b[0]))
print(type(b))
```
