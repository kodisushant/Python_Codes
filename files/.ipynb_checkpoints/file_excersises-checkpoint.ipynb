```python
''' 
Q5	Marks Report Generator	Easy

The file marks.txt contains student names and marks (one per line, separated by a tab).
Write a program that reads this file and creates a new file result.txt where each line shows:

Ravi    : 85  -> PASS
Priya   : 92  -> PASS
Arjun   : 78  -> PASS
Meena   : 88  -> PASS
Kiran   : 74  -> PASS

•	A student PASSES if marks >= 40, otherwise FAIL.
•	Read from marks.txt, write the result to result.txt.

'''
with open("marks.txt","r") as f:
    s=f.readline()
    new = []
    while s!="": 
        if int(s.split()[1]) >= 40: 
            s=s[:-1]+" -> PASS"
        else : 
            s=s[:-1]+" -> FAIL"
        new.append(s)
        s=f.readline() 
with open("results.txt","w") as f:
    f.write("\n".join(new))
```

```python
''' 
Q6	Find and Replace in a File	Medium

Write a program that:

1.	Reads the entire content of a text file into a string.
2.	Replaces every occurrence of a given word (e.g., "Python") with another word (e.g., "Java").
3.	Writes the modified content back to the SAME file.

•	The replacement must be case-insensitive (replace 'python', 'Python', 'PYTHON' all).
•	Print how many replacements were made.

'''
with open("story.txt", "r") as f:
    s = f.readlines()

replaced = []
count = 0

for line in s:
    words = line.split(" ")          # split by single space to better preserve structure
    for j in range(len(words)):
        if words[j].lower().strip() == "python":
            words[j] = "Java"
            count += 1
    replaced.append(words)           # append every line, not just matched ones

combine_replaced = ""
for line in replaced:
    combine_replaced += " ".join(line)   # no extra \n — split lines already have \n at end
print(combine_replaced)
with open("story.txt", "w") as f:
    f.write(combine_replaced)

print(f"Replacements made: {count}")
```

```python
''' 
Q8	Remove Blank Lines	Medium

Write a program that reads a text file that contains several blank lines mixed in,
and writes a clean version of the file (without any blank lines) to a new file called cleaned.txt.

•	A blank line is one that is empty or contains only whitespace.
•	Print how many blank lines were removed.

Hint: Check if line.strip() == '' to identify blank lines

'''
with open("messy.txt","r") as f:
    c=0
    clean = []
    line = f.readline()
    while line!="": 
        if line.strip() == '': 
            c+=1
        else:
            clean.append(line.strip())
        line = f.readline() 
print(clean)
print(c)
with open("cleaned.txt","w") as f: 
    f.write("\n".join(clean))
```

```python
'''
Q10	Copy a File (Manual Implementation)	Medium

Without using shutil.copy(), write your own file copy function:

def copy_file(source, destination):
    # Your code here
    pass

•	Read from source in binary mode ('rb') and write to destination in binary mode ('wb').
•	This makes it work for both text files AND image/audio files.
•	After copying, verify by comparing file sizes using os.path.getsize().

Hint: Binary mode ('rb'/'wb') reads/writes raw bytes — works for any file type.
'''
import os
def copy_file(source , destination): 
    with open(source,"rb") as f , open(destination,"wb") as out: 
        content = f.read() 
        out.write(content)
s=input("Enter the file to be copied:")
d=input("Enter copy file name:")
copy_file(s,d)
print(f"Source size      : {os.path.getsize(s)} bytes")
print(f"Destination size : {os.path.getsize(d)} bytes")  # ✅ Fix 1: 'd' not 'b'

if os.path.getsize(s) == os.path.getsize(d):
    print("✅ Copy verified — sizes match!")
else:
    print("❌ Something went wrong — sizes differ.")
```

```python
''' 
Q11	Safe File Reader with Exception Handling	Medium

Write a function safe_read(filename) that:

4.	Tries to open and read the file.
5.	If the file is not found, prints a friendly error message and returns None.
6.	If a PermissionError occurs, prints an appropriate message and returns None.
7.	If any other error occurs, prints the error type and message.
8.	Always prints 'Operation complete.' whether or not an error occurred.

safe_read('notes.txt')      # normal case
safe_read('missing.txt')    # FileNotFoundError
safe_read('/root/secret')   # PermissionError

Hint: Use try / except FileNotFoundError / except PermissionError / except Exception / finally.

'''
def safe_read(filename):
    try: 
        with open(filename,"r") as f:
            content = f.read()
            print(content)
            return content
    except FileNotFoundError : 
        print("Error: File not found — check the filename and try again.")
        return None
    except  PermissionError :
        print("Error: Permission denied — you don't have access to this file.")
        return None
    except Exception as e: 
        print(f"Unexpected Error:{type(e).__name__}-{e}")
        return None
    finally:
        print("operation complete")
# file_name = input("enter file name:")
safe_read(None)
safe_read("marks.txt")
safe_read("12")
safe_read("/root/secret")
```

```python
''' 
Q12	Write a Log File	Medium

Write a function log_event(message) that appends a timestamped entry to a file called app.log:

log_event('Application started')
log_event('User Priya logged in')
log_event('File processed successfully')

Expected Output:
[2024-07-15 09:30:01] Application started
[2024-07-15 09:30:05] User Priya logged in
[2024-07-15 09:30:07] File processed successfully

Hint: Use datetime.now().strftime('%Y-%m-%d %H:%M:%S') to get a formatted timestamp. Open in 'a' mode

'''
from datetime import datetime
def log_event(message): 
    with open("app.log","a") as f:
        timestamp = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
        f.write(f"[{timestamp}] {message}\n")
log_event('Application started')
log_event('User Priya logged in')
log_event('File processed successfully')
```

```python
''' 
Q13	Student Record Manager (File-Based)	Medium

Build a simple console-based student record system that stores data in students.txt. 
Each line in the file stores one student in the format:

RollNo,Name,Marks

Implement these four operations:

9.	add_student(roll, name, marks) — appends a new student record
10.	view_all() — reads and displays all records in a table format
11.	search_student(roll) — finds and prints the student with the given roll number
12.	delete_student(roll) — removes the student with the given roll number

Hint: For delete, read all lines, filter out the target, then rewrite the file in 'w' mode.

'''
def add_student(roll, name,marks): 
    with open("students.txt","a") as f:
        f.write(f"{roll},{name},{marks}\n")
def view_all(): 
    with open("students.txt","r") as f: 
        content = f.read()
        print(content)
def search_student(roll): 
    with open("students.txt","r") as f: 
        line = f.readline()
        while line!="": 
            if line.split(",")[0]==roll: 
                print(line)
                return line
            line=f.readline()
        print("not found")
def delete_student(roll): 
    with open("students.txt","r") as f: 
        line = f.readline()
        back_up=[]
        while line!="": 
            if line.split(",")[0]!=roll: 
                back_up.append(line)
            line=f.readline()
    with open("students.txt","w") as f:
        f.writelines(back_up)
add_student("101", "Priya", "88")
add_student("102", "Arjun", "76")
add_student("103", "Sneha", "91")
view_all()
search_student("102")
delete_student("102")
print("\nAfter deleting 102:")
view_all()
```

```python
''' 
Q14	Word Frequency Counter	Medium

Write a program that reads a text file and counts how many times each word appears.
Write the top 10 most frequent words (with counts) to a file called word_freq.txt,
sorted by frequency (highest first):

Expected Output:
the       : 42
is        : 31
python    : 28
...       : ...

•	Ignore punctuation and treat words case-insensitively.
•	Ignore common stop words: 'the', 'a', 'an', 'is', 'in', 'of', 'and', 'to'.

Hint: Use a dictionary to count, then sorted(d.items(), key=lambda x: x[1], reverse=True).

'''
import re

freq = {}
stop_words =['the', 'a', 'an', 'is', 'in', 'of', 'and', 'to']
with open("story_for_freq.txt", "r") as f:
    for line in f:                                      
        words = re.split(r"[\s.,;!?\"\'()\-:;]+", line.lower())  
        for word in words:
            if word == "" or word in stop_words:                             
                continue
            freq[word] = freq.get(word, 0) + 1        


sorted_freq = sorted(freq.items(), key=lambda x: x[1], reverse=True)


for word, count in sorted_freq[:10]:
    print(f"{word:<12}: {count}")

with open("word_freq.txt", "w") as f:
    for word, count in sorted_freq[:10]:
        f.write(f"{word:<12}: {count}\n")
```

```python
''' 
Q15	File-Based To-Do List	Hard

Build a command-line To-Do list application that persists tasks in a file called todo.txt.
Support the following commands:

python todo.py add  'Buy groceries'
python todo.py add  'Submit assignment'
python todo.py list
python todo.py done 1
python todo.py list

Expected Output:
# After two adds:
1. [ ] Buy groceries
2. [ ] Submit assignment

# After done 1:
1. [x] Buy groceries
2. [ ] Submit assignment

Hint: Store each task as a line: '[ ] task text' or '[x] task text'. Use sys.argv for CLI args.

'''

# todo.py
import sys

FILE = "todo.txt"

def load_tasks():
    try:
        with open(FILE, "r") as f:
            return [line.rstrip("\n") for line in f if line.strip()]
    except FileNotFoundError:
        return []   # no file yet = no tasks

def save_tasks(tasks):
    with open(FILE, "w") as f:
        f.write("\n".join(tasks) + "\n" if tasks else "")

def add_task(text):
    tasks = load_tasks()
    tasks.append(f"[ ] {text}")
    save_tasks(tasks)
    print(f"Added: '{text}'")

def list_tasks():
    tasks = load_tasks()
    if not tasks:
        print("No tasks yet!")
        return
    for i, task in enumerate(tasks, start=1):   # start=1 → 1,2,3 not 0,1,2
        print(f"{i}. {task}")

def mark_done(num):
    tasks = load_tasks()
    index = num - 1                             # user gives 1-based, list is 0-based

    if index < 0 or index >= len(tasks):
        print(f"No task number {num}.")
        return

    task_text = tasks[index].replace("[ ]", "[x]", 1)   # only replace first occurrence
    tasks[index] = task_text
    save_tasks(tasks)
    print(f"Marked task {num} as done.")

# ─── CLI Entry Point ───────────────────────────────────────────────
if len(sys.argv) < 2:
    print("Usage: python todo.py [add|list|done] [args]")
    sys.exit()

command = sys.argv[1]       # sys.argv[0] = "todo.py" (script name)
                            # sys.argv[1] = "add" / "list" / "done"
                            # sys.argv[2] = task text or number

if command == "add":
    if len(sys.argv) < 3:
        print("Usage: python todo.py add 'task text'")
    else:
        add_task(sys.argv[2])

elif command == "list":
    list_tasks()

elif command == "done":
    if len(sys.argv) < 3:
        print("Usage: python todo.py done <task number>")
    else:
        mark_done(int(sys.argv[2]))

else:
    print(f"Unknown command: '{command}'. Use add / list / done.")
```

```python
''' 
Q16	Encrypt and Decrypt a File (Caesar Cipher)	Hard

Write two functions:

•	encrypt_file(input_file, output_file, shift) — reads a text file and writes an encrypted version by shifting each letter by shift positions.
•	decrypt_file(input_file, output_file, shift) — reverses the encryption.

encrypt_file('message.txt', 'secret.txt', 3)
decrypt_file('secret.txt', 'decoded.txt', 3)
# decoded.txt should match original message.txt

•	Only shift alphabetic characters; leave spaces, digits, punctuation unchanged.
•	Preserve upper/lowercase.

Hint: Use ord() and chr(). For uppercase: chr((ord(c) - ord('A') + shift) % 26 + ord('A')).

'''
def caesar_cipher(text, shift):
    result = ""
    for c in text:
        if c.isupper():
            result += chr((ord(c) - ord('A') + shift) % 26 + ord('A'))
        elif c.islower():
            result += chr((ord(c) - ord('a') + shift) % 26 + ord('a'))
        else:
            result += c          
    return result

def encrypt_file(input_file, output_file, shift):
    with open(input_file, "r") as f:
        content = f.read()
    encrypted = caesar_cipher(content, shift)
    with open(output_file, "w") as f:
        f.write(encrypted)
    print(f"Encrypted → '{output_file}'")

def decrypt_file(input_file, output_file, shift):
    with open(input_file, "r") as f:
        content = f.read()
    decrypted = caesar_cipher(content, -shift)  
    with open(output_file, "w") as f:
        f.write(decrypted)
    print(f"Decrypted → '{output_file}'")


encrypt_file("message.txt", "secret.txt", 3)
decrypt_file("secret.txt", "decoded.txt", 3)
```
