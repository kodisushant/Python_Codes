```python
with open("marks.txt","r") as n:
    edited = []
    i = n.readline()
    while i != "": 
        if i.split()[-1] == 'f' and int(i.split()[-2]) >60:
            i=i[:-1]+" GOLD"
            print(i)
        elif i.split()[-1] == 'f' and int(i.split()[-2]) <60:
            i=i[:-1]+" BRONZE"
        elif i.split()[-1] == 'm' and int(i.split()[-2]) >60:
            i=i[:-1]+" SILVER"
        elif i.split()[-1] == 'm' and int(i.split()[-2]) <60:
            i=i[:-1]+" STEEL"
        edited.append(i)

print(edited)
```
