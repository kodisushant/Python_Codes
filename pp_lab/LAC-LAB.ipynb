```python
## mulitiplication of 2 matrix:
r1=int(input("Enter row size : "))
c1=int(input("Enter col size : "))
mat1=[]
for i in range (r1) :
    row=[]
    for j in range(c1) :
        row.append(int(input()))
    mat1.append(row)
print("Matrix1 : ",)
print("[",end=" ")
for i in range (r1) :
    for j in range(c1) :
        print(mat1[i][j],end=" ")
    print()
print("]",end=" ")
r2=int(input("Enter row size : "))
c2=int(input("Enter col size : "))
mat2=[]
for i in range (r2) :
    row=[]
    for j in range(c2) :
        row.append(int(input()))
    mat2.append(row)
print("Matrix2 : ")
print("[",end=" ")
for i in range (r2) :
    for j in range(c2) :
        print(mat2[i][j],end=" ")
    print()
print("]",end=" ")
print("\n")
if c1==r2 :
    mat3=[]
    for i in range (r1) :
        row=[]
        for j in range(c2) :
            x=0
            for k in range(r2) :
                x=x+(mat1[i][k]*mat2[k][j])
            row.append(x)
        mat3.append(row)
    print("[",end=" ")
    for i in range(r1) :
        for j in range(c2) :
            print(mat3[i][j],end=" ")
        print()
    print("]",end=" ")
else :
    print("Matrix multiplication is not possible ")
```

```python
## trace of matrix
r=int(input('Enter row:'))
c=int(input('Enter columns'))
mat=[]
if r==c:
    for i in range(r):
        row=[]
        for j in range(c):
            row.append(int(input()))
        m.append(row)
    print('Matrix:\n[')
    trace_sum=0
    for i in range(r):
        for j in range(c):
            print(m[i][j],end=" " )
        print()
    print(']')
    print('trace element:',end=" ")
    for i in range(r):
        for j in range(c):
            if i==j:
                print(m[i][j],end=" ")
                trace_sum += m[i][j]
    print(f"\nTrace of matrix (sum): {trace_sum}")
else:
    print('it is not a sqaure matrix')
```

```python
#to calculate diagonalization of a matrix
import sympy as sp
import numpy as np
A=sp.Matrix([[1,1,1],[0,2,1],[-4,4,3]])
print('Matrix=\n',A)
eigen=A.eigenvects()
print('Eigen values ,NO.of times repeateed , eigen vectors:\n',eigen)
A=np.array([[1,1,1],[0,2,1],[-4,4,3]])
B=np.array([[-1/2,-1,1],[1,1,0],[1,1,1]])
p=B.T
#model matrix
print('Model Matrix\n',p)
pinverse=np.linalg.inv(p)
print('Inverse of P\n',pinverse)
D=pinverse@A@p
print('Diagonal Matrix\n',D)
```

```python
# program on angle between the two surfaces at given point
from sympy import*
import numpy as np
def gradient(f,points):
    dx=diff(f,x)
    dy=diff(f,y)
    dz=diff(f,z)
    grad=np.array([dx.subs(points),dy.subs(points),dz.subs(points)])
    return grad
def gradient_norm(grad):
    s=0
    for v in grad:
        s=s+v**2
    return sqrt(s)
x,y,z=symbols('x,y,z')
s1=x**2+y**2+z**2-9
s2=x**2+y**2-z-2
points={x:2,y:-1,z:2}
grad1=gradient(s1,points)
print('Gradiant of surface 1 =',grad1)
grad2=gradient(s2,points)
print('Gradinat of surface 2 =',grad2)
n1=gradient_norm(grad1)                                   #output: Gradiant of surface 1:[4 -2 4]
                                                        # Gradiant of surface 2:[4 -2 -1]
                                                        # theta = acos(8*sqrt(21))
n2=gradient_norm(grad2)
Costheta =np.dot(grad1,grad2)/(n1*n2)
print('Theta=',acos(Costheta))
```
