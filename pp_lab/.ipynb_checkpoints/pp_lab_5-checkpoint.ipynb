```python
#To solve Application Problem in System of Equations 
''' 
A lady bought 12 apples and 6 bananas for $102 on a specific day.The next day , she bought 10 apples and 10 bananas for $110.
assuming the proce of bananas didn't change in two days , what is the price of one apple and one banana 
12a + 6b = 102 , 10a + 10b = 110
'''
import numpy as np 
A = np.array([[12,6],[10,10]])
B = np.array([102,110])
X = np.linalg.solve(A,B)
print("The price of one apple and one banana is",X)
```

```python
# 2x+3y+4z=5 , x+2y+3z=4 , 3x-4y+7z=10 
import numpy as np 
A=np.array([[2,3,4],[1,2,3],[3,-4,7]])
B = np.array([5,4,10])
X= np.linalg.solve(A,B)
print("Solution:",X)
```

```python
import sympy as sp 
mat = sp.Matrix([[1,3,5],
                 [0,2,7],
                 [0,0,9]])
print(mat)
eigen = mat.eigenvects() 
for i in range(3): 
    print("\neigen value",eigen[i][0])
    print("number of times eigen value repeated",eigen[i][1])
    print("eigen vector",eigen[i][2])
```
