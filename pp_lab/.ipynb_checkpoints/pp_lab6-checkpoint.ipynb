```python
#to find eigen values and eigen vectors 
import numpy as np 
#create matrix 
matrix = np.array([[1,-1,3],
                   [1,1,6],
                   [3,8,9]])
print("Square matrix:\n",matrix)

#Calculate eigenvalues and eigenvectors (Use Numpy's linalg.eig )
eigenvalues , eigenvectors = np.linalg.eig(matrix)

print("\nEigen values:\n",eigenvalues)

#In numpy we get normalized eigen vectors 
print("\neigen Vectors:\n",eigenvectors)
```

```python
# Matrix Diagonalization: D = P_inv * A * P
import numpy as np

# Define the original matrix
A = np.array([
    [ 1, 1, 1],
    [ 0, 2, 1],
    [-4, 4, 3]
], dtype=float)

print("Original Matrix A:\n", A)

# Step 1: Compute eigenvalues and normalized eigenvectors directly
eigenvalues, P = np.linalg.eig(A)
# P columns are the normalized eigenvectors automatically

print("\nEigenvalues:\n", eigenvalues)
print("\nModal Matrix P (normalized eigenvectors as columns):\n", P)

# Step 2: Compute inverse of modal matrix
P_inv = np.linalg.inv(P)
print("\nInverse of P:\n", P_inv)

# Step 3: Compute diagonal matrix D = P_inv * A * P
D = P_inv @ A @ P
D_clean = np.round(D.real, decimals=6)  # round off floating point noise
print("\nDiagonal Matrix D:\n", D_clean)

# Step 4: Verify — reconstruct A from D and check
A_reconstructed = P @ D @ P_inv
print("\nReconstructed A (verify):\n", np.round(A_reconstructed.real, decimals=6))
print("\nDiagonalization successful?", np.allclose(A, A_reconstructed.real))
```

```python
# Matrix Diagonalization: D = P_inv * A * P

import sympy as sp
import numpy as np

# Define the original matrix
A = sp.Matrix([
    [1, 1,  1],
    [0, 2,  1],
    [-4, 4, 3]
])
print("Original Matrix A:\n", A)

# Step 1: Compute eigenvalues and eigenvectors using SymPy
# eigenvects() returns list of (eigenvalue, multiplicity, [eigenvector])
eigen_data = A.eigenvects()
print("\nEigenvalues, Multiplicity, Eigenvectors:\n", eigen_data)

# Step 2: Build the modal matrix P (columns = eigenvectors)
# Extract eigenvectors and stack them as columns
eigenvectors = [vec for val, mult, vecs in eigen_data for vec in vecs]
modal_matrix = sp.Matrix.hstack(*eigenvectors)  # horizontal stack = column matrix
print("\nModal Matrix P (SymPy):\n", modal_matrix)

# Convert to NumPy for numerical computation
A_numpy = np.array(A.tolist(), dtype=float)
P = np.array(modal_matrix.tolist(), dtype=float ) # modal matrix as numpy array

print("\nModal Matrix P (NumPy):\n", P)

# Step 3: Compute the inverse of the modal matrix
P_inv = np.linalg.inv(P)
print("\nInverse of Modal Matrix (P_inv):\n", P_inv)

# Step 4: Compute the diagonal matrix D = P_inv * A * P
D = P_inv @ A_numpy @ P
print("\nDiagonal Matrix D:\n", np.round(D, decimals=6))  # round to remove floating point noise
```

```python
#To calculate Diagonalization of a Matrix

import sympy as sp
import numpy as np

A = sp.Matrix([[1,1,1],[0,2,1],[-4,4,3]])
print("Matrix = \n",A)

eigen = A.eigenvects()
print("Eigen Values , No of times repeated , Eigen Vectors\n",eigen)

A = np.array([[1,1,1],[0,2,1],[-4,4,3]])

B = np.array([[-1/2,-1,1],[1,1,0],[1,1,1]])

P = B.T
print("Modal Matrix\n",P)

Pinverse = np.linalg.inv(P)
print("Inverse of P\n",Pinverse)

D = Pinverse @ A @ P
print("Diagonal Matrix\n",D)
```
