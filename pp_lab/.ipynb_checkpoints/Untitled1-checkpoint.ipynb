```python
from sympy import * 
import numpy as np 
def gradient(f,points): 
    dx = diff(f,x)
    dy = diff(f,y)
    dz = diff(f,z)
    grad = np.array([dx.subs(points),dy.subs(points),dz.subs(points)])
    return grad 
def gradient_norm(grad):
    s=0 
    for v in grad: 
        s+=v**2 
    return sqrt(s)
x,y,z = symbols("x,y,z")
s1 = x**2 + y**2 + z**2 - 9 
s2  = x**2 + y**2 - z - 2 
points = {x:2,y:-1,z:2}
grad1 = gradient(s1,points)
print("Gradient of surface 1 =",grad1)
grad2 = gradient(s2,points)
print("Gradient of surface 2 =",grad2)
n1 = gradient_norm(grad1)
n2 = gradient_norm(grad2)

costheta = np.dot(grad1,grad2)/(n1*n2) 
print("Theta=",acos(costheta))
```
