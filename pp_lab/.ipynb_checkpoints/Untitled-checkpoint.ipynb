```python
from sympy import * 
x,y = symbols("x,y")
function = x**2*y**3 + 12*y**4
print("Partial derivative wrt x=",diff(function,x))
print("Partial derivative wrt y=",diff(function,y))
print("2nd Order Partial Derivative wrt x=",diff(function,x,x))
print("2nd Order Partial Derivative wrt y=",diff(function,y,y))
print("2nd Order Partial Derivative wrt y and x=",diff(function,y,x))
print("2nd Order Partial Derivative wrt x and y=",diff(function,x,y))
```

```python
from sympy import *
x,y,z = symbols("x,y,z")
print("Give the functions u,v,w in separate lines")
u = input("Give the function u:")
v = input("Give the function v:")
w = input("Give the function w:")
a1,a2,a3 = diff(u,x) , diff(u,y),diff(u,z)
b1,b2,b3 = diff(v,x) , diff(v,y),diff(v,z)
c1,c2,c3 = diff(w,x) , diff(w,y),diff(w,z)
result = Matrix([[a1,a2,a3],[b1,b2,b3],[c1,c2,c3]])
print(result) 
print("The Jacobian of u,v,w with respect to x,y,z is",result.det())
```

```python
from sympy import * 
x,y = symbols("x,y")
#constant limits 
print("Double integral with constant limits:",integrate(x*y,(x,0,1),(y,0,1)))
#variable limits
print("Double integral with constant limits:",integrate(x*y,(x,0,y),(y,0,1)))
```

```python
from sympy import * 
x , y,z = symbols("x,y,z")
# Constant Limits 
print("Triple integral with constant limits:",integrate(x*y*z,(x,0,1),(y,0,1),(z,0,1)))
#Variable limits 
print("Triple integral with variable limits:",integrate(x*y*z,(z,0,x+y),(y,0,1-x)))
```

```python
#Program on Gradient for given function 
from sympy import * 
from numpy import array 
x,y,z = symbols("x,y,z") 
f=x*y*z 
dx = diff(f,x)
dy = diff(f,y)
dz = diff(f,z)
grad = array([dx,dy,dz])
print("Gradient =",grad)
```

```python
from sympy import * 
from numpy import array 
x,y,z = symbols("x,y,z")
variables = symbols("x,y,z")
# given vector f1*i + f2*j + f3*k 
f1 = x**2 
f2 = y**3 
f3 = x*z 
F = array([f1,f2,f3])
s= 0 
for i in range(3): 
    s=s+diff(F[i],variables[i])
print("Divergence = ",s)
```

```python
#Curl for given function 
from sympy import *
import numpy as np 
from numpy import array 
x,y,z = symbols("x,y,z")
# Given vectors f1*i + f2*j + f3*k 
f1 = x*y*z 
f2 = 3*x**2*y 
f3 = (x*z**2)-(y**2*z)
point = {x:1,y:-1,z:1}
i =( diff(f3,y)-diff(f2,z) )
j= (diff(f3,x)-diff(f1,z))
k= diff(f2,x) - diff(f1,y) 
curl = array([i.subs(point),j.subs(point),z.subs(point)])
print("Curl=",curl)
```

```python
#Progame on volume integral 
from sympy import * 
from numpy import array 
x,y,z = symbols("x,y,z")
f1 = 2*x*z
f2 = -x 
f3 = y**2 
F = array([f1,f2,f3])
s = [] 
for i in range(3):
    s.append(integrate(F[i],(z,x**2,4),(x,0,2),(y,0,6)))
print("Volume Integral =",s)
```
