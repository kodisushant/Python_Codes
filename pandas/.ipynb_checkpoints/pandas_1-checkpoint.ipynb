```python
import pandas as pd 
import numpy as np 
import random
random.seed(0)
days = ["mon","tue","wed","thur","fri","sat","sun"]
x = pd.Series( [np.round(np.random.random()*100, 2) for i in range(7)],index=days )
print(x)
mintemp = pd.Series([random.randint(5,35) for i in range(7)],index = days)
maxtemp = pd.Series([random.randint(mintemp[i],45) for i in range(7)],index = days)

print(mintemp)
print(maxtemp)
print() 
```

```python
print("thursday",mintemp["thur"])
print("Wetttest day",days[np.argmax(x)],np.max(x))
x["mon"]=100.0
print(x[x==100.0])
print(mintemp.idxmin())
print()
print("min temperature when rainfall was > 50")
print( mintemp[x>50] )
print()
print( maxtemp[x>50])
```

```python
print( mintemp[x.idxmin()])
print( maxtemp[x.idxmin()])
```

```python
print( maxtemp[x>50] )
```
