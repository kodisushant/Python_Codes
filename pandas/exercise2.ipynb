```python
'''

Exercise 1 Hospital Patient Temperature Analysis Using a 2D NumPy Array
Problem Statement
A hospital records the body temperatures of patients for three consecutive days.
The temperature data is stored in a 2D NumPy array where:
•	Each row represents a patient 
•	Each column represents temperature recorded on a specific day 
•	Given 2D Array
    [98.6, 99.1, 100.2]
    [101.0, 102.5, 101.8]
    [97.9, 98.4, 98.7]
    [100.5, 101.2, 102.0]
days = ["Day1", "Day2", "Day3"]
patients = ["Patient_A", "Patient_B", "Patient_C", "Patient_D"]

Create a Pandas DataFrame from the NumPy array and perform the following tasks:
1.	Display the complete patient temperature DataFrame. 
2.	Display patients whose Day2 temperature is greater than 100°F. 
3.	Calculate the total temperature readings of each patient and store them in a new column called Total_Temp. 
4.	Calculate the average temperature of each patient and store it in a new column called Average_Temp. 
5.	Find the patient with the highest average temperature. 
6.	Add a new column called Health_Status: 
o	"Critical" → if average temperature is greater than or equal to 100°F 
o	"Normal" → otherwise 
7.	Calculate the temperature difference between Day3 and Day1 and store it in a new column called Temp_Difference. 
'''
import numpy as np 
import pandas as pd 
temp = np.array([[98.6, 99.1, 100.2],
    [101.0, 102.5, 101.8],
    [97.9, 98.4, 98.7],
    [100.5, 101.2, 102.0]])
days = np.array(["Day1", "Day2", "Day3"])
patients = np.array(["Patient_A", "Patient_B", "Patient_C", "Patient_D"]) 
d = {days[i]:temp[:,i] for i in range(len(days))}
# print(d)
df = pd.DataFrame(d,index=patients)
print(df)
print()
print(df[df["Day2"]>100])
print()
df["Total"] = df["Day1"]+df["Day2"]+df["Day3"]
print(df)
df["avg temp"] = round(df["Total"]/3,2) 
print()
print(df)
print() 
print(df["avg temp"].idxmax())

df["Health Status"] = df["avg temp"].apply(lambda x:"Critical" if x > 100 else "Normal")
print(df)
print()
df["Temp difference"] = df["Day3"] - df["Day1"]
df
```

```python
''' 
Exercise 2 : Online Food Delivery Orders Analysis Using Pandas
Problem Statement
A food delivery company tracks the number of weekly orders received from different restaurants.
The order data for three weeks is stored in a 2D NumPy array.
Numpy array
    [450, 520, 610],
    [300, 410, 450],
    [700, 820, 910],
    [500, 540, 600]
# Column names and restaurant names
weeks = ["Week1", "Week2", "Week3"]
restaurants = ["Pizza Hub", "Burger Town", "Spicy Kitchen", "Food Palace"]
Create a Pandas DataFrame from the given data and perform the following tasks:
1.	Display the complete restaurant orders DataFrame. 
2.	Display restaurants with Week2 orders greater than 500. 
3.	Calculate the total orders for each restaurant and store them in a new column called Total_Orders. 
4.	Find the restaurant with the highest total orders. 
5.	Add a new column called Performance: 
o	"Top" → if total orders are greater than or equal to 1500 
o	"Average" → otherwise 
6.	Calculate the growth in orders from Week1 to Week3 and store it in a new column called Growth. 
'''
import numpy as np 
import pandas as pd
data = np.array([[450, 520, 610],
    [300, 410, 450],
    [700, 820, 910],
    [500, 540, 600]])
weeks = np.array(["Week1", "Week2", "Week3"])
restaurants = np.array(["Pizza Hub", "Burger Town", "Spicy Kitchen", "Food Palace"])
orders = pd.DataFrame(data,index=restaurants,columns=weeks)
print(orders)
print()
print(orders[orders["Week2"]>500 ])
print()
orders["Total"] = orders[weeks].sum(axis=1)
print(orders)
print() 
print(orders["Total"].idxmax())
print() 
orders["Perfomance"] = np.where(orders["Total"]>=1500,"Top","Average")
print(orders)
print()
orders["Growth"] = orders["Week3"] - orders["Week1"]
print(orders)
```

```python
''' 
Q20. End-to-end pipeline — read, process, save

Write a complete program that does the following in sequence:

a) Read students.csv (Name, RollNo, Maths, Physics, City).

b) Add a Total column (Maths + Physics).

c) Add an Average column (Total / 2, rounded to 1 decimal place).

d) Add a Result column: if Average > 80 then Pass, else Fail.

e) Save the full processed DataFrame to processed_results.csv without index.

f) From the saved file, read back only the students who Passed and print them.

g) Print how many students Passed and how many Failed.

Expected output:

# Passing students read back from file:
    Name  RollNo  Maths  Physics  Total  Average Result
0   Ravi     101     85       76    161     80.5   Pass
1  Priya     102     92       88    180     90.0   Pass
3  Meena     104     95       90    185     92.5   Pass
4  Kiran     105     88       79    167     83.5   Pass

Pass     4
Fail     1
Name: Result, dtype: int64


---

'''
print() 
```

```python
import numpy as np 
import pandas as pd  

df = pd.read_csv("student2.csv")
df["Total"] = df["Maths"] + df["Physics"] 
df["Average"] = round(df["Total"]/2,1)
df["Result"] = np.where(df["Average"]>80.0,"Pass","Fail")
df.to_csv("processed_results.csv",index=False) 
new_df = pd.read_csv("processed_results.csv")
print(new_df[new_df["Result"]=="Pass"])
p = sum( new_df["Result"]=="Pass" )
f = sum(new_df["Result"] == "Fail")
print("Pass  ",p)
print("Fail  ",f)

print(new_df["Result"].value_counts())
```
